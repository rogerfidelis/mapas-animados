```python
import pandas as pd
import plotly.graph_objects as go
from gerar_mapa import BASE

# =========================
# CONFIGURAÇÕES
# =========================

EMPRESAS = ["bram", "starnav", "cbo"]

FUNDO = "#0B1F3A"
BRANCO = "#FFFFFF"

for EMPRESA in EMPRESAS:

    ARQUIVO = BASE / "dados" / f"SURVEY_{EMPRESA}_D2.xlsx"

    # =========================
    # LÊ TODAS AS ABAS
    # =========================

    abas = pd.read_excel(
        ARQUIVO,
        sheet_name=None
    )

    # =========================
    # PASTA DE SAÍDA
    # =========================

    PASTA = BASE / "graficos" / "distancia" / EMPRESA
    PASTA.mkdir(
        parents=True,
        exist_ok=True
    )

    # =========================
    # UM GRÁFICO PARA CADA ABA
    # =========================

    for nome_aba, df in abas.items():

        # Ignora abas vazias
        if df.empty:
            continue

        # =========================
        # TRATAMENTO DA DATA
        # =========================

        df["data_reportada"] = pd.to_datetime(
            df["data_reportada"]
        )

        # Label para exibição
        df["data_label"] = df["data_reportada"].dt.strftime(
            "%d/%m/%Y %H:%M"
        )

        # =========================
        # CRIAÇÃO DO GRÁFICO
        # =========================

        fig = go.Figure()

        fig.add_bar(
            x=df["data_label"],
            y=df["distancia_km"],
            marker_color="steelblue"
        )

        # =========================
        # EIXO X
        # =========================

        fig.update_xaxes(
            type="category",
            tickangle=90,

            # Letras
            color=BRANCO,

            # Linha do eixo
            linecolor=BRANCO,

            # Marcas
            tickcolor=BRANCO,

            # Grid
            gridcolor="rgba(255,255,255,0.15)"
        )

        # =========================
        # EIXO Y
        # =========================

        fig.update_yaxes(

            # Letras
            color=BRANCO,

            # Linha do eixo
            linecolor=BRANCO,

            # Marcas
            tickcolor=BRANCO,

            # Grid
            gridcolor="rgba(255,255,255,0.15)"
        )

        # =========================
        # LAYOUT
        # =========================

        fig.update_layout(

            # Fundo externo
            paper_bgcolor=FUNDO,

            # Fundo da área do gráfico
            plot_bgcolor=FUNDO,

            # Todas as fontes
            font=dict(
                color=BRANCO,
                family="Arial"
            ),

            # Título
            title=dict(
                text="Delta de deslocamento por intervalo de coleta",
                font=dict(
                    color=BRANCO,
                    size=20
                )
            ),

            # Títulos dos eixos
            xaxis_title=dict(
                text="Data da coleta",
                font=dict(
                    color=BRANCO
                )
            ),

            yaxis_title=dict(
                text="Distância (km)",
                font=dict(
                    color=BRANCO
                )
            ),

            height=550,
            width=1200,

            margin=dict(
                b=180
            )
        )

        # =========================
        # SALVA HTML
        # =========================

        arquivo_saida = (
            nome_aba
            .lower()
            .replace(" ", "_")
            + ".html"
        )

        fig.write_html(
            PASTA / arquivo_saida,
            include_plotlyjs="cdn",
            full_html=False
        )

        print(f"{arquivo_saida} criado.")
        
```
