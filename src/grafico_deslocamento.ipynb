```python
import pandas as pd
import plotly.graph_objects as go
from gerar_mapa import BASE

# =========================
# ARQUIVO DE DADOS
# =========================

ARQUIVO = BASE / "dados" / "SURVEY_starnav_D1.xlsx"

# Lê todas as abas
abas = pd.read_excel(ARQUIVO, sheet_name=None)

# Pasta onde serão salvos os gráficos
PASTA = BASE / "graficos" / "distancia" / "starnav"
PASTA.mkdir(parents=True, exist_ok=True)

# =========================
# GERA UM GRÁFICO PARA CADA ABA
# =========================

for nome_aba, df in abas.items():

    # ignora abas vazias
    if df.empty:
        continue

    # converte a data
    #df["data_consulta"] = pd.to_datetime(df["data_consulta"])

    #fig = go.Figure()

   # fig.add_bar(
   #     x=df["data_consulta"],
   #     y=df["distancia_km"],
   #     marker_color="steelblue"
   # )

   # fig.update_xaxes(
   # tickformat="%d/%m/%Y\n%H:%M",
   # tickangle=90,
   # tickmode="array",
   # tickvals=df["data_consulta"]
   # )

    df["data_consulta"] = pd.to_datetime(df["data_consulta"])

    # Cria uma coluna apenas para exibição
    df["data_label"] = df["data_consulta"].dt.strftime("%d/%m/%Y %H:%M")

    fig = go.Figure()

    fig.add_bar(
        x=df["data_label"],
        y=df["distancia_km"],
        marker_color="steelblue"
    )

    fig.update_xaxes(
        type="category",
        tickangle=90
    )


    fig.update_layout(
    title="Delta de deslocamento por intervalo de coleta",
    xaxis_title="Data da coleta",
    yaxis_title="Distância (km)",
    height=550,
    width=1200,
    margin=dict(b=180)
    )

    arquivo_saida = nome_aba.lower().replace(" ", "_") + ".html"

    fig.write_html(
        PASTA / arquivo_saida,
        include_plotlyjs="cdn",
        full_html=False
    )

    print(f"{arquivo_saida} criado.")
    
```
