```python
from pathlib import Path
import pandas as pd
from gerar_mapa import LINK_HOME
# ============================================================
# CONFIGURAÇÃO
# ============================================================
EMPRESAS = ["bram", "starnav", "cbo"]

for EMPRESA in EMPRESAS:
    

    # ============================================================
    # CAMINHOS
    # ============================================================
    BASE = Path.cwd().parent

    TEMPLATE = BASE / "templates" / "dashboard_rev2.html"

    saida= BASE / "dashboards"/ EMPRESA
    
    saida.mkdir(
    parents=True,
    exist_ok=True
    )

    # ============================================================
    # DADOS DAS EMBARCAÇÕES
    # ============================================================
    vessels = pd.read_excel(BASE / "dados" / "SURVEY_VESSELS" / "SURVEY_VESSELS.xlsx")

    # ============================================================
    # GERAÇÃO DOS DASHBOARDS
    # ============================================================

    for nome in vessels[EMPRESA].dropna():
        
        def normalizar_nome(nome):
            return (
                str(nome)
                .strip()
                .lower()
                .replace("/", "_")
                .replace("\\", "_")
                .replace(" ", "_")
            )

        #arquivo = nome.lower().replace(" ", "_")
        arquivo = normalizar_nome(nome)

        mapa = f"../../mapas/{EMPRESA}/{arquivo}.html"
        grafico = f"../../graficos/status_navegacao/{EMPRESA}/{arquivo}.html"
        grafico_deslocamento = (f"../../graficos/distancia/{EMPRESA}/{arquivo}.html")
        logo = (f"../../imagens/logo_ofi.png")
        atendimento = f"../../graficos/campo_atendimento/{EMPRESA}/{arquivo}.html"
        acumulada = f"../../graficos/distancia_acumulada/{EMPRESA}/{arquivo}.html"
        gantt= f"../../graficos/gantt_atendimento/{EMPRESA}/{arquivo}.html"

        # -------------------------
        # Abre o template
        # -------------------------

        with open(TEMPLATE, encoding="utf-8") as f:
            html = f.read()

        html = html.replace("{{NOME}}", nome)
        html = html.replace("{{MAPA}}", mapa)
        html = html.replace("{{STATUS}}", grafico)
        html = html.replace("{{DESLOCAMENTO}}", grafico_deslocamento)
        html = html.replace("{{HOME}}", LINK_HOME)
        html = html.replace("{{LOGO}}", logo)
        html = html.replace("{{ATENDIMENTO}}", atendimento)
        html = html.replace("{{ACUMULADA}}", acumulada)
        html = html.replace("{{GANTT}}", gantt)

        with open(saida / f"{arquivo}.html", "w", encoding="utf-8") as f:
            f.write(html)
```
