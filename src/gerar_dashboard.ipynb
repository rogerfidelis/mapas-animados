```python
from pathlib import Path
import pandas as pd

BASE = Path.cwd().parent

TEMPLATE = BASE / "templates" / "dashboard.html"



vessels = pd.read_excel(BASE / "dados" / "SURVEY_VESSELS" / "SURVEY_VESSELS.xlsx")

#saida= BASE / "dashboards"/ "cbo"
#saida= BASE / "dashboards"/ "bram"
saida= BASE / "dashboards"/ "starnav"

for nome in vessels["starnav"].dropna():

    arquivo = nome.lower().replace(" ", "_")

    mapa = f"../../mapas/starnav/{arquivo}.html"
    grafico = f"../../graficos/status_navegacao/starnav/{arquivo}.html"
    grafico_deslocamento = (f"../../graficos/distancia/starnav/{arquivo}.html")

    with open(TEMPLATE, encoding="utf-8") as f:
        html = f.read()

    html = html.replace("{{NOME}}", nome)
    html = html.replace("{{MAPA}}", mapa)
    html = html.replace("{{STATUS}}", grafico)
    html = html.replace("{{DESLOCAMENTO}}", grafico_deslocamento)

    with open(saida / f"{arquivo}.html", "w", encoding="utf-8") as f:
        f.write(html)
        
```
