```python
from openpyxl import load_workbook
import pandas as pd
from gerar_mapa import BASE

EMPRESA = "bram"

# Arquivo origem (A)
ARQUIVO_ORIGEM = BASE / "dados" / "SURVEY_VESSELS"/"SURVEY_VESSELS_20260728_manha.xlsx"
origem = pd.read_excel(ARQUIVO_ORIGEM, sheet_name= EMPRESA)

# =========================
# CABEÇALHOS DA PLANILHA A
# =========================

cabecalhos_A = {}

for i, coluna in enumerate(origem.columns):
    cabecalhos_A[str(coluna).strip()] = i

print(cabecalhos_A)

# Arquivo destino (B)
ARQUIVO_DESTINO = BASE / "dados" /f"SURVEY_{EMPRESA.upper()}.xlsx"

wb = load_workbook(ARQUIVO_DESTINO)

# =========================
# TESTE
# =========================
for _, linha in origem.iterrows():

    # Nome da aba de destino
    aba = str(linha.iloc[0]).strip().upper()

    print(f"Procurando aba: {aba}")

    # Verifica se a aba existe
    if aba not in wb.sheetnames:
        print(f"Aba '{aba}' não encontrada!")
        continue

    # Abre a aba
    ws = wb[aba]

    print(f"Aba encontrada: {ws.title}")

    # Apenas para testar a primeira embarcação
    
    # =========================
    # CABEÇALHOS DA ABA DESTINO
    # =========================

    cabecalhos_B = {}

    for coluna in range(1, ws.max_column + 1):

        nome_coluna = ws.cell(row=1, column=coluna).value

        if nome_coluna is not None:
            cabecalhos_B[str(nome_coluna).strip()] = coluna


    mapeamento= {
                     "data": "data_consulta",
                     "hora": "hora_consulta",
                     "lat": "lat_a",
                     "lon": "lon_a",
                     "data reportada": "data_reportada",
                     "hora reportada": "hora_reportada",
                     "status": "status"
                 }
    
    ws = wb[aba]

    linha_destino = ws.max_row + 1

    #print(f"Próxima linha livre: {linha_destino}")
    
    for coluna_origem, coluna_destino in mapeamento.items():

        # Verifica se a coluna existe nas duas planilhas
        if coluna_origem not in cabecalhos_A:
            print(f"Coluna '{coluna_origem}' não existe na planilha A.")
            continue

        if coluna_destino not in cabecalhos_B:
            print(f"Coluna '{coluna_destino}' não existe na aba {aba}.")
            continue

        # Copia o valor
        valor = linha[coluna_origem]

        # Escreve na planilha destino
        ws.cell(
            row=linha_destino,
            column=cabecalhos_B[coluna_destino]
        ).value = valor

wb.save(ARQUIVO_DESTINO)
```
