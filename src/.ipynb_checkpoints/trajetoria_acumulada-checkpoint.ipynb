```python
import pandas as pd
import plotly.graph_objects as go
from gerar_mapa import BASE

# =========================
# ARQUIVO DE DADOS
# =========================

EMPRESAS= ["bram", "starnav", "cbo"]

for EMPRESA in EMPRESAS:

    ARQUIVO = BASE / "dados" / f"SURVEY_{EMPRESA}_D2.xlsx"

    # Lê todas as abas
    abas = pd.read_excel(
        ARQUIVO,
        sheet_name=None
    )

    # =========================
    # PERCORRE TODAS AS EMBARCAÇÕES
    # =========================

    for nome_aba, df in abas.items():

        print(f"Processando: {nome_aba}")

        if "data_consulta" not in df.columns:
            print(f"Aba {nome_aba} ignorada.")
            continue

        if "distancia_km" not in df.columns:
            print(f"Aba {nome_aba} não possui distancia_km.")
            continue

        # =========================
        # PREPARAÇÃO
        # =========================

        df["data_reportada"] = pd.to_datetime(
            df["data_reportada"],
            dayfirst=True
        )

        df["distancia_km"] = pd.to_numeric(
            df["distancia_km"],
            errors="coerce"
        )

        df = df.dropna(
            subset=["data_reportada", "distancia_km"]
        )

        df = df.sort_values("data_reportada")

        # =========================
        # DISTÂNCIA ACUMULADA
        # =========================

        df["distancia_acumulada_km"] = (
            df["distancia_km"].cumsum()
        )

        # =========================
        # GRÁFICO
        # =========================

        fig = go.Figure()

        fig.add_trace(
            go.Scatter(
                x=df["data_consulta"],
                y=df["distancia_acumulada_km"],
                mode="lines+markers",
                name="Distância acumulada"
            )
        )

        fig.update_xaxes(
            tickformat="%d/%m/%Y %H:%M",
            tickangle=90,
            tickmode="array",
            tickvals=df["data_reportada"]
        )

        fig.update_layout(
            title=f"Distância acumulada - {nome_aba}",
            xaxis_title="Data Reportada",
            yaxis_title="Distância acumulada (km)",
            height=550,
            hovermode="x unified"
        )

        # =========================
        # NOME DO ARQUIVO
        # =========================

        arquivo = (
            nome_aba
            .lower()
            .replace(" ", "_")
        )

        saida = ( BASE/ "graficos" / "distancia_acumulada" / EMPRESA)

        saida.mkdir(
            parents=True,
            exist_ok=True
        )

        fig.write_html(
            saida / f"{arquivo}.html",
            include_plotlyjs="cdn",
            full_html=False
        )

        print(
            f"Gráfico salvo: {saida / f'{arquivo}.html'}"
        )

    print("Todos os gráficos foram gerados.")
```
