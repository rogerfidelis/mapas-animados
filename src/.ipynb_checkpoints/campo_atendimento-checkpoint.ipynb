```python
import pandas as pd
import plotly.express as px
from gerar_mapa import BASE

# =========================
# ARQUIVO DE DADOS
# =========================

EMPRESAS = ["cbo", "bram", "starnav"]

for EMPRESA in EMPRESAS:

    ARQUIVO = BASE / "dados" / f"SURVEY_{EMPRESA}_D2.xlsx"

    # Lê todas as abas
    abas = pd.read_excel(ARQUIVO, sheet_name=None)

    # =========================
    # GERA UM GRÁFICO POR ABA
    # =========================

    for nome_aba, df in abas.items():

        # Ignora a aba de posições
        if nome_aba.lower() == "posicoes":
            continue

        print(f"Gerando gráfico: {nome_aba}")

        # Considera somente registros com evidência de proximidade suficiente
        df_atendimento = df[
            df["evidencia_proximidade"] >= 0.5
        ].copy()

        status = (
            df_atendimento["campo"]
            .dropna()
            .value_counts()
        )

        fig = px.pie(
            values=status.values,
            names=status.index,
            hole=0.60
        )

        fig.update_traces(
            textposition="inside",
            textinfo="percent+label"
        )

        fig.update_layout(
            template="plotly_white",
            showlegend=True,

            # =========================
            # TÍTULO
            # =========================

            title=dict(
                text=f"Atendimento - {nome_aba}",
                x=0.5,
                xanchor="center",
                y=0.95,
                yanchor="top"
            ),

            # =========================
            # LEGENDA ABAIXO DA ROSCA
            # =========================

            legend=dict(
                orientation="h",
                yanchor="top",
                y=-0.10,
                xanchor="center",
                x=0.5
            ),

            # =========================
            # ESPAÇAMENTO
            # =========================

            margin=dict(
                t=80,
                b=100,
                l=20,
                r=20
            )
        )

        nome_arquivo = (
            nome_aba.lower()
            .replace(" ", "_")
            .replace("á", "a")
            .replace("é", "e")
            .replace("í", "i")
            .replace("ó", "o")
            .replace("ú", "u")
            .replace("ã", "a")
            .replace("ç", "c")
        )

        fig.write_html(
            BASE / "graficos" / "campo_atendimento" / EMPRESA /
            f"{nome_arquivo}.html",
            include_plotlyjs="cdn",
            full_html=False
        )

    print("Gráficos gerados com sucesso!")
```
