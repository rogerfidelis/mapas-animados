```python
import pandas as pd
import plotly.express as px
from gerar_mapa import BASE
import numpy as np
import plotly.io as pio


# ============================================================
# CONFIGURAÇÕES
# ============================================================

EMPRESAS = ["cbo", "bram", "starnav"]

for EMPRESA in EMPRESAS:

    ARQUIVO = ( BASE  / "dados" / f"SURVEY_{EMPRESA}_D2.xlsx")

    # ============================================================
    # DIRETÓRIO DE SAÍDA
    # ============================================================

    PASTA_SAIDA = ( BASE / "graficos" / "gantt_atendimento" / EMPRESA)

    PASTA_SAIDA.mkdir(
        parents=True,
        exist_ok=True
    )


    # ============================================================
    # LEITURA DOS DADOS
    # ============================================================

    abas = pd.read_excel( ARQUIVO, sheet_name=None)


    # ============================================================
    # PROCESSAMENTO DAS ABAS
    # ============================================================

    for nome_aba, df in abas.items():

        # --------------------------------------------------------
        # Ignora aba de posições
        # --------------------------------------------------------

        if nome_aba.lower() == "posicoes":
            continue

        print(f"\n{'=' * 70}")
        print(f"Processando Gantt: {nome_aba}")
        print(f"{'=' * 70}")


        # ========================================================
        # PREPARAÇÃO DOS DADOS
        # ========================================================

        # --------------------------------------------------------
        # Verifica se as colunas necessárias existem
        # --------------------------------------------------------

        colunas_necessarias = [
            "evidencia_proximidade",
            "data_reportada",
            "unidade_proxima"
        ]

        colunas_faltantes = [
            coluna
            for coluna in colunas_necessarias
            if coluna not in df.columns
        ]

        if colunas_faltantes:

            print(
                f"Colunas ausentes na aba {nome_aba}: "
                f"{colunas_faltantes}"
            )

            continue


        # --------------------------------------------------------
        # Converte evidência para número
        # --------------------------------------------------------

        df["evidencia_proximidade"] = pd.to_numeric(
            df["evidencia_proximidade"],
            errors="coerce"
        )


        # --------------------------------------------------------
        # Converte data_reportada para datetime
        # --------------------------------------------------------

        df["data_reportada"] = pd.to_datetime(
            df["data_reportada"],
            dayfirst=True,
            errors="coerce"
        )


        # --------------------------------------------------------
        # Remove linhas sem data
        # --------------------------------------------------------

        df_base = (
            df
            .dropna(subset=["data_reportada"])
            .sort_values("data_reportada")
            .reset_index(drop=True)
        )


        # --------------------------------------------------------
        # Identifica evidência suficiente
        # --------------------------------------------------------

        df_base["atendimento"] = (
            df_base["evidencia_proximidade"] >= 0.5
        )


        # ========================================================
        # MOSTRA DATAFRAME BASE
        # ========================================================

        print("\nDATAFRAME BASE:")

        print(
            df_base[
                [
                    "data_reportada",
                    "evidencia_proximidade",
                    "unidade_proxima",
                    "atendimento"
                ]
            ].to_string(index=False)
        )


        # ========================================================
        # CONSTRUÇÃO DOS EVENTOS
        # ========================================================

        registros_gantt = []

        evento_atual = None


        # --------------------------------------------------------
        # Percorre todas as observações
        # --------------------------------------------------------

        for i in range(len(df_base)):

            linha = df_base.iloc[i]

            data = linha["data_reportada"]

            evidencia = linha["evidencia_proximidade"]

            unidade = linha["unidade_proxima"]


            # ====================================================
            # CASO 1
            # Evidência insuficiente
            # ====================================================

            if pd.isna(evidencia) or evidencia < 0.5:

                # ----------------------------------------------
                # Fecha evento que estava aberto
                # ----------------------------------------------

                if evento_atual is not None:

                    evento_atual["Fim"] = data

                    if evento_atual["Fim"] > evento_atual["Inicio"]:

                        registros_gantt.append(
                            evento_atual.copy()
                        )

                    evento_atual = None

                continue


            # ====================================================
            # Caso 2
            # Evidência suficiente, mas unidade inexistente
            # ====================================================

            if pd.isna(unidade):

                if evento_atual is not None:

                    evento_atual["Fim"] = data

                    if evento_atual["Fim"] > evento_atual["Inicio"]:

                        registros_gantt.append(
                            evento_atual.copy()
                        )

                    evento_atual = None

                continue


            # ====================================================
            # PRIMEIRO EVENTO
            # ====================================================

            if evento_atual is None:

                evento_atual = {

                    "Inicio": data,

                    "Fim": data,

                    "Origem": unidade,

                    "Destino": unidade,

                    "Tipo": "Parado",

                    "Atividade": unidade

                }

                continue


            # ====================================================
            # EXISTE EVENTO ABERTO
            # ====================================================

            origem_atual = evento_atual["Destino"]


            # ====================================================
            # CASO 3
            # Mesma unidade
            #
            # Continua o mesmo evento
            # ====================================================

            if unidade == origem_atual:

                evento_atual["Fim"] = data

                continue


            # ====================================================
            # CASO 4
            # Mudança de unidade
            #
            # Fecha o evento anterior e cria um evento de trânsito
            # ====================================================

            # ----------------------------------------------------
            # Fecha evento anterior
            # ----------------------------------------------------

            evento_atual["Fim"] = data

            if evento_atual["Fim"] > evento_atual["Inicio"]:

                registros_gantt.append(
                    evento_atual.copy()
                )


            # ----------------------------------------------------
            # Cria evento de trânsito
            #
            # O trânsito representa a mudança entre:
            #
            # unidade anterior → nova unidade
            # ----------------------------------------------------

            evento_atual = {

                "Inicio": data,

                "Fim": data,

                "Origem": origem_atual,

                "Destino": unidade,

                "Tipo": "Trânsito",

                "Atividade": (
                    f"{origem_atual} → {unidade}"
                )

            }


        # ========================================================
        # FECHA EVENTO FINAL
        # ========================================================

        if evento_atual is not None:

            ultima_data = df_base.iloc[-1]["data_reportada"]

            evento_atual["Fim"] = ultima_data

            if evento_atual["Fim"] > evento_atual["Inicio"]:

                registros_gantt.append(
                    evento_atual.copy()
                )


        # ========================================================
        # DATAFRAME DO GANTT
        # ========================================================

        df_gantt = pd.DataFrame(
            registros_gantt
        )


        # --------------------------------------------------------
        # Proteção contra DataFrame vazio
        # --------------------------------------------------------

        if df_gantt.empty:

            print(
                f"\nNenhum evento válido para gerar Gantt: "
                f"{nome_aba}"
            )

            continue


        # ========================================================
        # REMOVE INTERVALOS INVÁLIDOS
        # ========================================================

        df_gantt = df_gantt[
            df_gantt["Fim"] > df_gantt["Inicio"]
        ].reset_index(drop=True)


        # --------------------------------------------------------
        # Proteção
        # --------------------------------------------------------

        if df_gantt.empty:

            print(
                f"\nNenhum intervalo temporal válido: "
                f"{nome_aba}"
            )

            continue


        # ========================================================
        # DURAÇÃO
        # ========================================================

        df_gantt["Duracao"] = (
            df_gantt["Fim"]
            - df_gantt["Inicio"]
        )


        # ========================================================
        # IDENTIFICADOR ÚNICO DO EVENTO
        # ========================================================

        df_gantt["id_evento"] = np.arange(
            len(df_gantt)
        )


        # ========================================================
        # IDENTIFICADOR DO EIXO Y
        # ========================================================

        df_gantt["Atividade_Y"] = (

            df_gantt["id_evento"]
            .astype(str)

            + " - "

            + df_gantt["Atividade"]
        )


        # ========================================================
        # ORDEM DOS EVENTOS
        # ========================================================

        ordem_eventos = (
            df_gantt["Atividade_Y"]
            .tolist()
        )


        # ========================================================
        # MOSTRA DATAFRAME FINAL
        # ========================================================

        print("\nDATAFRAME FINAL DO GANTT:")

        print(
            df_gantt[
                [
                    "id_evento",
                    "Inicio",
                    "Fim",
                    "Duracao",
                    "Atividade",
                    "Atividade_Y",
                    "Tipo",
                    "Origem",
                    "Destino"
                ]
            ].to_string(index=False)
        )


        # ========================================================
        # CORES
        # ========================================================

        cores_tipo = {

            "Trânsito": "#1CA3EC",

            "Parado": "#D4AF37"

        }


        # ========================================================
        # GANTT
        # ========================================================

        fig = px.timeline(

            df_gantt,

            x_start="Inicio",

            x_end="Fim",

            y="Atividade_Y",

            color="Tipo",

            color_discrete_map=cores_tipo,

            hover_data={

                "Origem": True,

                "Destino": True,

                "Duracao": True,

                "Tipo": True,

                "Inicio": True,

                "Fim": True,

                "Atividade_Y": False

            },

            category_orders={

                "Atividade_Y": ordem_eventos

            }

        )


        # ========================================================
        # EIXO X
        # ========================================================

        fig.update_xaxes(

            title_text="Data / Hora",

            title_font=dict(
                size=22
            ),

            tickfont=dict(
                size=18
            ),

            tickformat="%d/%m %H:%M",

            dtick=6 * 60 * 60 * 1000

        )


        # ========================================================
        # EIXO Y
        # ========================================================

        fig.update_yaxes(

            title_text="Atividade",

            title_font=dict(
                size=22
            ),

            tickfont=dict(
                size=16
            )

        )


        # ========================================================
        # LAYOUT
        # ========================================================

        fig.update_layout(

            #width=1400,

            height=700,

            margin=dict(

                l=300,

                r=50,

                t=120,

                b=80

            ),

            font=dict(

                size=18,

                family="Arial"

            ),

            title=dict(

                text=f"""
                <b>Atendimento Offshore</b><br>
                <span style='font-size:18px'>{nome_aba}</span>
                """,

                font=dict(
                    size=22
                ),

                x=0.5,

                xanchor="center"

            ),

            legend=dict(

                title="Tipo de atendimento",

                title_font=dict(
                    size=16
                ),

                font=dict(
                    size=14
                )

            )

        )


        # ========================================================
        # BORDA DAS BARRAS
        # ========================================================

        fig.update_traces( marker_line_color="white", marker_line_width=1)


        # ========================================================
        # SALVAMENTO DO GANTT
        # ========================================================

        nome_arquivo = (
            str(nome_aba)
            .strip()
            .title()
            .replace("/", "_")
            .replace("\\", "_")
            .replace(" ", "_")
        )

        arquivo_saida = (PASTA_SAIDA/f"{nome_arquivo}.html")


        fig.write_html(
            arquivo_saida,
            full_html=True,
            include_plotlyjs=True,
            config={
                "responsive": True
            }
        )


        print( f"Gantt salvo em:\n{arquivo_saida}" )

```
