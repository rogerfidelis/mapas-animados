```python
import pandas as pd
import plotly.express as px
from gerar_mapa import BASE

# ============================================================
# CONFIGURAÇÕES
# ============================================================

ARQUIVOS = {
    "BRAM": BASE / "dados" / "SURVEY_BRAM_D2.xlsx",
    "CBO": BASE / "dados" / "SURVEY_CBO_D2.xlsx",
    "STARNAV": BASE / "dados" / "SURVEY_STARNAV_D2.xlsx"
}

# ============================================================
# PROCESSAMENTO
# ============================================================

for empresa, arquivo in ARQUIVOS.items():

    print("\n" + "=" * 70)
    print(f"EMPRESA: {empresa}")
    print("=" * 70)

    planilhas = pd.read_excel(
        arquivo,
        sheet_name=None
    )

    dados_empresa = []

    for nome_aba, df in planilhas.items():

        # Ignora abas que não possuem distância
        if "distancia_km" not in df.columns:
            continue

        # Converte para número
        distancia = pd.to_numeric(
            df["distancia_km"],
            errors="coerce"
        ).sum()

        dados_empresa.append({
            "embarcacao": nome_aba,
            "distancia_km": distancia
        })

        print(
            f"{nome_aba:<30} "
            f"{distancia:>12,.2f} km"
        )

    # ========================================================
    # DATAFRAME DA EMPRESA
    # ========================================================

    df_empresa = pd.DataFrame(dados_empresa)

    # Ordena da maior para a menor distância
    df_empresa = df_empresa.sort_values(
        "distancia_km",
        ascending=False
    )

    # ========================================================
    # GRÁFICO
    # ========================================================

    fig = px.bar(
        df_empresa,
        x="embarcacao",
        y="distancia_km",
        text="distancia_km",
        title=f"Distância Percorrida — {empresa}",
        labels={
            "embarcacao": "Embarcação",
            "distancia_km": "Distância total (km)"
        }
    )

    # ========================================================
    # CONFIGURAÇÃO DAS BARRAS
    # ========================================================

    fig.update_traces(

        texttemplate="%{text:,.0f} km",

        textposition="outside",

        hovertemplate=(
            "<b>%{x}</b><br>"
            "Distância total: %{y:,.2f} km"
            "<extra></extra>"
        )
    )

    # ========================================================
    # LAYOUT
    # ========================================================

    fig.update_layout(

        template="plotly_dark",

        title_x=0.5,

        height=600,

        xaxis_title="Embarcação",

        yaxis_title="Distância total percorrida (km)",

        xaxis={
            "categoryorder": "total descending"
        },

        margin=dict(
            l=80,
            r=40,
            t=90,
            b=120
        )
    )

    # Inclinação dos nomes das embarcações
    fig.update_xaxes(
        tickangle=-45
    )

    # ========================================================
    # ARQUIVO DE SAÍDA
    # ========================================================

    PASTA_SAIDA = BASE / "graficos" / "comparar_distancia"

    PASTA_SAIDA.mkdir(
        parents=True,
        exist_ok=True
    )

    arquivo_saida = (
        PASTA_SAIDA /
        f"comparar_distancia_{empresa.lower()}.html"
    )

    fig.write_html(
        arquivo_saida,
        include_plotlyjs="cdn"
    )

    print("-" * 70)

    print(
        f"Gráfico salvo em:\n"
        f"{arquivo_saida}"
    )

print("\n" + "=" * 70)
print("PROCESSAMENTO CONCLUÍDO")
print("=" * 70)
```
