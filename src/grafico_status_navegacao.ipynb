```python
import pandas as pd
import plotly.express as px
from gerar_mapa import BASE

# =========================
# ARQUIVO DE DADOS
# =========================

EMPRESA = "cbo"

ARQUIVO = BASE / "dados" / f"SURVEY_{EMPRESA}_D2.xlsx"

# Lê todas as abas
abas = pd.read_excel(ARQUIVO, sheet_name=None)

# =========================
# GERA UM GRÁFICO POR ABA
# =========================

for nome_aba, df in abas.items():

    # Ignora a aba de posições
    if nome_aba.lower() == "posicoes":
        continue

    print(f"Gerando gráfico: {nome_aba}")

    status = (
        df["status de navegação"]
        .dropna()
        .value_counts()
    )

    fig = px.pie(
        values=status.values,
        names=status.index,
        hole=0.60,
        title=f"Status de Navegação - {nome_aba}"
    )

    # =========================
    # CONFIGURAÇÃO DA ROSCA
    # =========================

    fig.update_traces(
        textposition="inside",
        textinfo="percent+label",
        domain=dict(
            x=[0, 1],
            y=[0.30, 0.90]
        )
    )

    # =========================
    # CONFIGURAÇÃO DO LAYOUT
    # =========================

    fig.update_layout(

        template="plotly_white",

        # Título
        title=dict(
            x=0.5,
            xanchor="center",
            y=0.97,
            yanchor="top"
        ),

        # Legenda abaixo da rosca
        legend=dict(
            orientation="h",
            yanchor="top",
            y=0.22,
            xanchor="center",
            x=0.5
        ),

        # Espaço interno do gráfico
        margin=dict(
            l=20,
            r=20,
            t=70,
            b=20
        ),

        showlegend=True
    )

    # =========================
    # NOME DO ARQUIVO
    # =========================

    nome_arquivo = (
        nome_aba.lower()
        .replace(" ", "_")
        .replace("á", "a")
        .replace("é", "e")
        .replace("í", "i")
        .replace("ó", "o")
        .replace("ú", "u")
        .replace("ã", "a")
        .replace("ç", "c")
    )

    fig.write_html(
        BASE / "graficos" / "status_navegacao" / EMPRESA /
        f"{nome_arquivo}.html",
        include_plotlyjs="cdn",
        full_html=False
    )

print("Gráficos gerados com sucesso!")
```
