```python
import pandas as pd
import plotly.express as px
from gerar_mapa import BASE

# =========================
# CONFIGURAÇÃO VISUAL
# =========================

FUNDO = "#0B1F3A"
BRANCO = "#FFFFFF"

# =========================
# ARQUIVO DE DADOS
# =========================

EMPRESAS = ["cbo", "bram", "starnav"]

for EMPRESA in EMPRESAS:

    ARQUIVO = BASE / "dados" / f"SURVEY_{EMPRESA}_D2.xlsx"

    # Lê todas as abas
    abas = pd.read_excel(
        ARQUIVO,
        sheet_name=None
    )

    # =========================
    # GERA UM GRÁFICO POR ABA
    # =========================

    for nome_aba, df in abas.items():

        # Ignora a aba de posições
        if nome_aba.lower() == "posicoes":
            continue

        print(f"Gerando gráfico: {nome_aba}")

        # =========================
        # FILTRO DE PROXIMIDADE
        # =========================

        df_atendimento = df[
            df["evidencia_proximidade"] >= 0.5
        ].copy()

        # =========================
        # CONTAGEM POR CAMPO
        # =========================

        status = (
            df_atendimento["campo"]
            .dropna()
            .value_counts()
        )

        # =========================
        # GRÁFICO
        # =========================

        fig = px.pie(
            values=status.values,
            names=status.index,
            hole=0.60
        )

        # =========================
        # TEXTO DA ROSCA
        # =========================

        fig.update_traces(

            # Mostra SOMENTE o percentual
            textposition="inside",
            textinfo="percent",

            textfont=dict(
                color=BRANCO,
                size=14
            ),

            # =========================
            # TOOLTIP
            # =========================

            hovertemplate=(
                "<b>%{label}</b><br>"
                "Atendimentos: %{value}<br>"
                "Percentual: %{percent}"
                "<extra></extra>"
            )
        )

        # =========================
        # LAYOUT
        # =========================

        fig.update_layout(

            # Fundo
            paper_bgcolor=FUNDO,
            plot_bgcolor=FUNDO,

            # Fonte geral
            font=dict(
                color=BRANCO,
                family="Arial"
            ),

            showlegend=True,

            # =========================
            # TÍTULO
            # =========================

            title=dict(
                text=f"Atendimento<br><b>{nome_aba}</b>",
                x=0.5,
                xanchor="center",
                y=0.95,
                yanchor="top",

                font=dict(
                    color=BRANCO,
                    size=18
                )
            ),

            # =========================
            # LEGENDA
            # =========================

            legend=dict(
                orientation="h",
                yanchor="top",
                y=-0.10,
                xanchor="center",
                x=0.5,

                font=dict(
                    color=BRANCO,
                    size=12
                )
            ),

            # =========================
            # ESPAÇAMENTO
            # =========================

            margin=dict(
                t=80,
                b=100,
                l=20,
                r=20
            )
        )

        # =========================
        # NOME DO ARQUIVO
        # =========================

        nome_arquivo = (
            nome_aba.lower()
            .replace(" ", "_")
            .replace("á", "a")
            .replace("é", "e")
            .replace("í", "i")
            .replace("ó", "o")
            .replace("ú", "u")
            .replace("ã", "a")
            .replace("ç", "c")
        )

        # =========================
        # EXPORTAÇÃO
        # =========================

        fig.write_html(
            BASE
            / "graficos"
            / "campo_atendimento"
            / EMPRESA
            / f"{nome_arquivo}.html",
            include_plotlyjs="cdn",
            full_html=False
        )

    print("Gráficos gerados com sucesso!")
```
